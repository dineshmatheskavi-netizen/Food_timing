```python
def inventory_tracker():
    stock_data = ["25", "5", "0", "abc", "-3", "12", "2"]

    healthy = 0
    critical = 0
    low = 0
    out_of_stock = 0
    invalid = 0

    for value in stock_data:
      try:
        stock=int(value)

        if stock < 0:
          print(value,"Invalid Stock")
          invalid += 1

        elif stock == 0:
          print(value,"out of stock")
          out_of_stock += 1

        elif stock <= 5:
          print(value,"Critical stock")
          critical += 1

        elif stock <=10:
          print(value,"Low stock")
          low += 1

        else:
          print(value,"Healthy stock")
          healthy += 1

      except ValueError:
          print(value,"Invaldi numeric value")
          invalid += 1

    print("\n--- Inventory Report ---")
    print("Healthy stock:", healthy)
    print("Critical stock:", critical)
    print("Low stock:", low)
    print("Out of stock:", out_of_stock)
    print("Invalid records:", invalid)

    print("\n--- Reorder Report ---")

    for value in stock_data:
      try:
          stock = int(value)

          if stock > 0 and stock <= 5:
            print("Reorder the stock",value)

      except ValueError:
        continue

inventory_tracker()
```
