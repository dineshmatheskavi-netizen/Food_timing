```python
def order_report():
    orders = [
        {"id": "101", "qty": "2", "price": "500"},
        {"id": "102", "qty": "0", "price": "300"},
        {"id": "103", "qty": "abc", "price": "700"},
        {"id": "104", "qty": "5", "price": "N/A"},
        {"id": "105", "qty": "-2", "price": "400"},
        {"id": "106", "qty": "3", "price": "1200"},
        {"id": "107", "qty": "4", "price": "-50"}
    ]

    total_revenue = 0
    valid_orders = 0
    invalid_orders = 0
    high_value = 0
    regular_orders = 0

    for order in orders:
        try:
            order_id = order["id"]
            qty = float(order["qty"])
            price = float(order["price"])

            if qty <= 0:
                print("Order", order_id, "-> Invalid quantity")
                invalid_orders += 1
                continue

            elif price <= 0:
                print("Order", order_id, "-> Invalid price")
                invalid_orders += 1
                continue

            revenue = qty * price
            total_revenue += revenue
            valid_orders += 1

            if revenue >= 2000:
                print("Order", order_id, "-> High-value order")
                high_value += 1
            else:
                print("Order", order_id, "-> Regular order")
                regular_orders += 1

            print("Revenue: ₹", revenue)

        except ValueError:
            print("Order", order["id"], "-> Invalid numeric data")
            invalid_orders += 1

    print("\n--- Business Report ---")
    print("Valid orders:", valid_orders)
    print("Invalid orders:", invalid_orders)
    print("High-value orders:", high_value)
    print("Regular orders:", regular_orders)
    print("Total revenue: ₹", total_revenue)

    if valid_orders > 0:
        average_revenue = total_revenue / valid_orders
        print("Average revenue per valid order: ₹", average_revenue)
    else:
        print("No valid orders available.")


order_report()
```
