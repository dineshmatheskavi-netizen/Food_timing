```python
def food_timing():
  try:
    food_intake_time = int(input("wating time for breakfast (5 to 8): "))
    for time in range(5,9):
      if food_intake_time == time:
        if food_intake_time <=6:
          print("Great!! u eat in good moring time")
          print("Have ur food and go peacefully")
        elif food_intake_time <=7:
          print("Nice! Good time for food")
          print("Eat ontime and ur cab is one the way")
        else:
          print("You are late. your cab is gone")
        break
    else:
      print("Please enter a time between 5 and 8.")

  except:
     print("Enter the valid and restricted food time")
```
